## **Ingesting S3 parquet to databricks delta managed table**

```python
s3_source_path = "s3a://hari-data-engineering-workspace/shopstream/raw/parquet/2019-Oct/"


df_raw = spark.read.format("parquet").load(s3_source_path)



df_raw.write.format("delta").mode("overwrite").saveAsTable("shopstream.raw.raw_events")
```
