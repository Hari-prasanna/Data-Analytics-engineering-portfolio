```sql
-- Creating data mask function on user_session column

CREATE OR REPLACE FUNCTION shopstream.gold.mask_user_session_token(session_id STRING)

RETURN
CASE 
  WHEN current_user() = 'hariprasannaravichandran@gmail.com' THEN session_id 
  ELSE CONCAT(LEFT(session_id, 5), '*****')
END;


-- Applying the Mask to the fact table


ALTER TABLE shopstream.gold.fct_session_funnel
ALTER COLUMN user_session SET MASK shopstream.gold.mask_user_session_token;
```
